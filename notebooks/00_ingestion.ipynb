```python
from minio import Minio
from minio.error import S3Error

client = Minio(
    "minio:9000",
    access_key = "admin",
    secret_key = "password123",
    secure = False
)
BUCKET = "rental-observatory"

if not client.bucket_exists(BUCKET):
    client.make_bucket(BUCKET)
    print(f"Bucket '{BUCKET}' made")
else:
    print(f"Bucket '{BUCKET}' already exists")
```

```python
files_to_upload = {
    "listings_NY.csv.gz":                       "../data/raw/listings_NY.csv.gz",
    "calendar_NY.csv.gz":                       "../data/raw/calendar_NY.csv.gz",
    "Zip_zori_uc_sfrcondomfr_sm_month.csv":  "../data/raw/Zip_zori_uc_sfrcondomfr_sm_month.csv",
    "ACSDT5Y2024.B19013-Data.csv":           "../data/raw/ACSDT5Y2024.B19013-Data.csv",
    "ACSDT5Y2024.B01003-Data.csv":           "../data/raw/ACSDT5Y2024.B01003-Data.csv",
}
```

```python
import os
for filename, local_path in files_to_upload.items():
    object_name = f"bronze/{filename}" #destination path inside MinIO

    if not os.path.exists(local_path):
        print(f" File not found: {local_path}")
        continue
    print(f" Uploading {filename}")
    client.fput_object(BUCKET, object_name,local_path)
    print(f"{object_name} uploaded")

print("Ingestion complete")
```
